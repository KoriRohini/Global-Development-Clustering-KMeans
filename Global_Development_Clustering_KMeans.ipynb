```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df =pd.read_csv("/content/Country-data.csv")
df
```

```python
print("First 5 record")
print(df.head)
```

```python
X=df.drop("country",axis=1)
X
```

```python
scaler =StandardScaler()
X_scaled =scaler.fit_transform(X)
print(X_scaled[:5])
```

```python
X_scaled=pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)
```

```python
KMeans = KMeans(n_clusters=3, random_state=42, n_init=10)
```

```python
KMeans.fit(X_scaled)
```

```python
df['Cluster']=KMeans.labels_
print("\nCountries with Cluster:")
print(df[['country','Cluster']].head(15))
```

```python
print("\nNumber of countries int each cluster:")
print(df['Cluster'].value_counts().sort_index())
```

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=3)
model.fit(X_scaled)
```

```python
import pandas as pd
new_Sample = pd.DataFrame(
    [[10, 40, 7, 35, 30000, 5, 75, 2, 15000]],
    columns=X.columns
)


new_Sample_scaled = pd.DataFrame(
    scaler.transform(new_Sample),
    columns=X.columns
)


new_cluster = model.predict(new_Sample_scaled)
```

```python
print("\nNew Country Sample:")
print(new_Sample)

print("\nNew Country belongs to Cluster:",
      new_cluster[0])
```
